```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Actividad 1

```python
#variable dependiente
alt=[17, 15, 14, 12, 11, 6, 4]
#variable independiente
tiempo=[8, 14, 22, 27, 33, 50, 60]
```

Actividad 2
crear dataFrame

```python
df=pd.DataFrame({
    "altura": alt,
    "tiempo": tiempo
})
```

```python
df["x*y"] = df["tiempo"] * df["altura"]
df["x^2"] = df["tiempo"] ** 2
df["y^2"] = df["altura"] ** 2
```

Mostrar tabla

```python
print("\n Tabla de datos\n")
print(df)
n = len(df)
```

Actividad 3 calcular sumatorias

```python
sum_x=df["tiempo"].sum()
sum_y=df["altura"].sum()
sum_xy=df["x*y"].sum()
sum_x2=df["x^2"].sum()
sum_y2=df["y^2"].sum()
```

```python
print("\n sumatorias \n")
print("suma x:", sum_x)
print("suma y:", sum_y)
print("suma xy:", sum_xy)
print("suma x^2", sum_x2)
print("suma y^2:", sum_y2)
```

Actividad 4 regresion lineal

Calcular:
Pendiente b,
Ordenada al origen a

```python
#formula de b
b=((n*sum_xy)-(sum_x*sum_y))/((n*sum_x2)-(sum_x**2))
#formula de a
a=((sum_y*sum_x2)-(sum_xy*sum_x))/((n*sum_x2)-(sum_x**2))
```

```python
print("\n recta de regresion \n")
print(f"y = {a:.2f}+{b:.2f}x")
```

Actividad 5 calcula coeficiente

```python
#formula de r
r=((n*sum_xy)-(sum_x*sum_y))/np.sqrt(((n*sum_x2)-(sum_x**2))*((n*sum_y2)-(sum_y**2)))
```

```python
print("\n Correlacion \n")
print(f"r = {r:.4f}")
```

```python
#formula de r2
r2=r**2
```

```python
print(f"r^2={r2:.4f}")
```

```python
#interpretaciones
if r > 0:
    print("\n la correlacion es positiva.")
elif r < 0:
    print("\n la correlacion es negativa.")
else:
    print("\n no existe.")

print(f"el modelo explica aproximadamente el {r2*100:.2f}% de la variabilidad.")
```

Actividad 6 grafico

```python
x=np.array(tiempo)
y=a+b*x

plt.scatter(tiempo, alt)
plt.plot(x, y)

plt.title("regresion lineal")
plt.xlabel("Tiempo")
plt.ylabel("Altura")

plt.show()
```
